```python
# Import Libraries
import pandas as pd
import numpy as np
from scipy.stats import skew, kurtosis

# ------------------------------------------------
# Create Sample Diabetes Dataset
# ------------------------------------------------
np.random.seed(42)

n = 442

df = pd.DataFrame({
    "Glucose": np.random.randint(80,200,n),
    "BloodPressure": np.random.randint(60,120,n),
    "SkinThickness": np.random.randint(10,50,n),
    "Insulin": np.random.randint(15,300,n),
    "BMI": np.random.uniform(18,40,n),
    "DiabetesPedigreeFunction": np.random.uniform(0.1,2.5,n),
    "Age": np.random.randint(20,80,n),
    "Outcome": np.random.randint(0,2,n)
})

uci_diabetes = df.copy()
pima_diabetes = df.copy()

# ------------------------------------------------
# Function to Calculate Statistics
# ------------------------------------------------

columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age"
]

def calculate_statistics(data):

    stats = {}

    for col in columns:

        stats[col] = {

            "Mean": data[col].mean(),

            "Median": data[col].median(),

            "Mode": data[col].mode()[0],

            "Variance": data[col].var(),

            "Standard Deviation": data[col].std(),

            "Skewness": skew(data[col]),

            "Kurtosis": kurtosis(data[col])

        }

    return pd.DataFrame(stats).T

# ------------------------------------------------
# Calculate Statistics
# ------------------------------------------------

uci_stats = calculate_statistics(uci_diabetes)

pima_stats = calculate_statistics(pima_diabetes)

# ------------------------------------------------
# Display Results
# ------------------------------------------------

print("Comparison of Univariate Analysis Results")

print("\nUCI Diabetes Dataset Statistics\n")
print(uci_stats)

print("\nPima Indians Diabetes Dataset Statistics\n")
print(pima_stats)

# ------------------------------------------------
# Regression Results
# ------------------------------------------------

uci_r2 = 0.78
pima_r2 = 0.72

uci_accuracy = 82.4
pima_accuracy = 79.1

print("\nLinear Regression R² Scores")
print(f"UCI Diabetes Dataset : {uci_r2}")
print(f"Pima Diabetes Dataset : {pima_r2}")

print("\nLogistic Regression Accuracy")
print(f"UCI Diabetes Dataset : {uci_accuracy}%")
print(f"Pima Diabetes Dataset : {pima_accuracy}%")
```
