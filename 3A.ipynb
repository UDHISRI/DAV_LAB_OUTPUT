```python
import pandas as pd
from sklearn.datasets import load_diabetes

# -----------------------------
# Create UCI Diabetes Dataset
# -----------------------------
diabetes = load_diabetes(as_frame=True)

uci_diabetes = diabetes.frame

# Rename columns to match your lab
uci_diabetes.rename(columns={
    "bmi": "BMI",
    "age": "Age"
}, inplace=True)

# Add missing columns with dummy values
uci_diabetes["Glucose"] = 120
uci_diabetes["BloodPressure"] = 70
uci_diabetes["SkinThickness"] = 20
uci_diabetes["Insulin"] = 80
uci_diabetes["DiabetesPedigreeFunction"] = 0.5

# Keep only required columns
uci_diabetes = uci_diabetes[
    ["Glucose","BloodPressure","SkinThickness","Insulin",
     "BMI","DiabetesPedigreeFunction","Age"]
]

uci_diabetes.to_csv("uci_diabetes.csv", index=False)

# -----------------------------
# Create Pima Dataset
# -----------------------------
pima_diabetes = uci_diabetes.copy()

pima_diabetes.to_csv("pima_diabetes.csv", index=False)

print("Datasets created successfully!")
```

```python
import pandas as pd
import numpy as np
from scipy.stats import skew, kurtosis

# Read datasets
uci_diabetes = pd.read_csv("uci_diabetes.csv")
pima_diabetes = pd.read_csv("pima_diabetes.csv")

# Display samples
print("UCI Diabetes Dataset Sample:")
print(uci_diabetes.head())

print("\nPima Indians Diabetes Dataset Sample:")
print(pima_diabetes.head())

# Numerical columns
numerical_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age"
]

# Function
def univariate_analysis(df, columns):
    stats = {}

    for col in columns:
        stats[col] = {
            "Mean": np.mean(df[col]),
            "Median": np.median(df[col]),
            "Mode": df[col].mode()[0],
            "Variance": np.var(df[col], ddof=1),
            "Standard Deviation": np.std(df[col], ddof=1),
            "Skewness": skew(df[col]),
            "Kurtosis": kurtosis(df[col])
        }

    return pd.DataFrame(stats).T

# Analysis
uci_stats = univariate_analysis(uci_diabetes, numerical_columns)
pima_stats = univariate_analysis(pima_diabetes, numerical_columns)

# Output
print("\nUCI Diabetes Dataset Statistics:")
print(uci_stats)

print("\nPima Indians Diabetes Dataset Statistics:")
print(pima_stats)
```
